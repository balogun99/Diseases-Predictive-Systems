# Import Libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

# Data Collection

### Kidney Disease Dataset

```python
# load the dataset
kidney_data = pd.read_csv("/Users/macbookpro/Desktop/ML & DL & GENAI Projects/Diseases Predictive System/Kidney Disease Predictive System/data/kidney_disease_dataset.csv")
kidney_data
```

# Data Preprocessing

```python
# first 5 rows of the dataset
kidney_data.head()
```

```python
# last 5 rows of the dataset
kidney_data.tail()
```

```python
# shape of the dataset
kidney_data.shape
```

```python
# check for missing values
kidney_data.isnull().sum()
```

```python
# check the distribution of the last column - Dialysis Needed
kidney_data['Dialysis_Needed'].value_counts()
```

## 0 ==> No Dialysis
## 1 ==> Dialysis Present

```python
# statistical measures of the dataset
kidney_data.describe()
```

```python
# separating the features and the target
X = kidney_data.drop(columns='Dialysis_Needed', axis=1)
y = kidney_data['Dialysis_Needed']
```

```python
print(X)
```

```python
print(y)
```

# Data Splitting

```python
# train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3,
                                                    random_state=42,
                                                    stratify=y)
```

```python
# shape of the training and testing
print(X.shape, y.shape, X_train.shape, X_test.shape)
```

# Model Training

```python
# train the model
classifier = svm.SVC(kernel='linear')
# fit the model
classifier.fit(X_train, y_train)
```

# Model Evaluation

# Accuracy Score

### Training Data

```python
# accuracy score on training data
X_train_pred = classifier.predict(X_train)
train_data_pred = accuracy_score(X_train_pred, y_train)
```

```python
# resullt
print("Accuracy Score on Training Data :", train_data_pred)
```

### Testing Data

```python
X_test_pred = classifier.predict(X_test)
test_data_pred = accuracy_score(X_test_pred, y_test)
```

```python
# result
print("Accuracy Score on Testing Data :", test_data_pred)
```

# Predictive System

```python
input_data  = (71,0.3,40.9,0,1,46.8,1622,1) # No Dialysis Present
# convert the input data into a numpy array
input_data_as_numpy_array = np.asarray(input_data)
# reshape the array for predicting 1 instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)
# model prediction
prediction = classifier.predict(input_data_reshaped)
print(prediction)
if (prediction[0] == 0):
    print("No Dialysis Present")
else:
    print("Dialysis Present")
```
