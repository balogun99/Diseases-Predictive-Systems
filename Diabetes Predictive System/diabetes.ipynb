# Data Splitting

```python
# separate features from the target
X = diabetes_data.drop(columns='Outcome', axis=1)
y = diabetes_data['Outcome']
```

```python
print(X)
```

```python
print(y)
```

```python
X.shape
```

```python
y.shape
```

# Data Standardization

```python
scaler = StandardScaler()
scaler.fit(X)
```

```python
standardized_data = scaler.transform(X)
print(standardized_data)
```

```python
X = standardized_data
y = diabetes_data['Outcome']
```

```python
print(X)
```

```python
print(y)
```

# Split Features & Target

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, stratify=y, random_state=42)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

# Model Training

```python
# train the model
classifier = svm.SVC(kernel='linear')
# fit the model
classifier.fit(X_train, y_train)
```

# Model Evaluation

## Accuracy Score

```python
# accuracy score on training data
X_train_pred = classifier.predict(X_train)
train_data_acc = accuracy_score(X_train_pred, y_train)
```

```python
# result on the training data
print("Accuracy Score on Training Data: ", train_data_acc)
```

```python
# accuracy score on testing data
X_test_pred = classifier.predict(X_test)
test_data_acc = accuracy_score(X_test_pred, y_test)
```

```python
# result on the testing data
print("Accuracy Score on Testing Data: ",test_data_acc)
```

# Predictive System

```python
# input_data = (4,110,92,0,0,37.6,0.191,30) # Non-diabetic
input_data = (6,148,72,35,0,33.6,0.627,50) # Diabetic

# change input data into a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array for predicting 1 instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = classifier.predict(input_data_reshaped)
print(prediction)

if (prediction[0] == 0):
    print("Non Diabetic Patient")
else:
    print("Diabetic Patient")
```
